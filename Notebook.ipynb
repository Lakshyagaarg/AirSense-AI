## Data Cleaning And Preprocessing

```python
df.drop(columns=['Date','City','Benzene','Toluene','Xylene'],inplace=True)
```

### Handling missing values

```python
df=pd.concat([df,df_station],ignore_index=True) #merging
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum() #checking the missing values
```

```python
# Filling the missing values by median value.
#For the numeric values only.
df.fillna(df.median(numeric_only=True),inplace=True)
```

```python
df.isnull().sum()
```

```python
# We will be removing rows that are not having the AQI labels for them.
df.dropna(subset=['AQI_Bucket'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

## Feature Preparation.

```python
df.head()
```

```python
aqi_mapping = {
    "Good": 0,
    "Satisfactory": 1,
    "Moderate": 2,
    "Poor": 3,
    "Very Poor": 4,
    "Severe": 5
}
df['AQI_Bucket'] = df['AQI_Bucket'].str.strip()
df['AQI_Bucket'] = df['AQI_Bucket'].map(aqi_mapping)
df.head()
```

```python
df['AQI_Bucket'].isnull().sum()
```

## Splitting of Data.

```python
x=df.drop(columns=['AQI_Bucket','AQI']) #Independent features i.e. pollutants

#Dependent features
y_reg=df['AQI']
y_classi=df['AQI_Bucket']
```

```python
x.shape, y_classi.shape, y_reg.shape
```

```python
#Splitting the data into training and test data.
# ratio is 70-30.
from sklearn.model_selection import train_test_split

X_train, X_test, y_train_reg, y_test_reg = train_test_split(
    x, y_reg, test_size=0.3, random_state=42
)

X_train2, X_test2, y_train_cls, y_test_cls = train_test_split(
    x, y_classi, test_size=0.3, random_state=42
)
```

```python
# Feature scalling so that each pollutant can be treated equally
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

X_train2 = scaler.fit_transform(X_train2)
X_test2 = scaler.transform(X_test2)
```

```python
# For Regression
X_train.shape, X_test.shape, y_train_reg.shape, y_test_reg.shape
```

```python
#for classification
X_train2.shape, X_test2.shape, y_train_cls.shape, y_test_cls.shape
```

## Model

### Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
```

```python
linear=LinearRegression() #it has multiple independent features.
```

```python
linear.fit(X_train,y_train_reg)
```

```python
y_pred_reg=linear.predict(X_test)
```

```python
# Performance Metrix
mae_lr = mean_absolute_error(y_test_reg, y_pred_reg)
mse_lr = mean_squared_error(y_test_reg, y_pred_reg)
rmse_lr = root_mean_squared_error(y_test_reg, y_pred_reg)
r2_lr = r2_score(y_test_reg, y_pred_reg)

print("Linear Regression Performance: ")
print("MAE :", mae_lr)
print("MSE :", mse_lr)
print("RMSE:", rmse_lr)
print("R2 Score:", r2_lr)
```

```python
plt.figure(figsize=(6,6))
sns.regplot(x=y_test_reg, y=y_pred_reg, scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.xlabel("Actual AQI")
plt.ylabel("Predicted AQI")
plt.title("Linear Regression: Actual vs Predicted AQI with Best-Fit Line")
plt.show()
```

```python
baseline_results = {
    "Model": "Linear Regression",
    "MAE": mae_lr,
    "RMSE": rmse_lr,
    "R2 Score": r2_lr,
    "MSE": mse_lr
}

import pandas as pd

result_df = pd.DataFrame([baseline_results])
result_df.to_csv(r"C:\Users\HP\OneDrive\Desktop\DataScience\AQI Project\Result_csv\model_results.csv", index=False)
```

```python
coefficients = pd.DataFrame({
    "Feature": x.columns,
    "Coefficient": linear.coef_
})

coefficients.sort_values(by="Coefficient", ascending=False)

# It is telling which pollutant affecting the AQI most in linear regression
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
```

```python
logistic.fit(X_train2,y_train_cls)
```

```python
y_pred_log=logistic.predict(X_test2)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
accuracy_logistic=accuracy_score(y_test_cls,y_pred_log)
precision_logistic=precision_score(y_test_cls,y_pred_log,average='weighted')
recall_logistic=recall_score(y_test_cls,y_pred_log,average='weighted')
f1_logistic=f1_score(y_test_cls,y_pred_log,average='weighted')
print("Logistic Regression Performance: ")
print("Accuracy :", accuracy_logistic)
print("Precision :", precision_logistic)
print("Recall :", recall_logistic)
print("F1 Score :", f1_logistic)
```

```python
cm=confusion_matrix(y_test_cls,y_pred_log)
plt.figure(figsize=(6,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression: Confusion Matrix")
plt.show()
```

```python
logistic_result={
    "Model":"Logistic Regression",
    "Accuracy":accuracy_logistic,
    "Precision":precision_logistic,
    "Recall":recall_logistic,
    "F1 Score":f1_logistic
}
```

```python
classi_result_df=pd.DataFrame([logistic_result])
```

### Random Forest Classifier
